# ЗАДАНИЕ: Консольное приложение для учета складских заявок на выдачу материалов

Необходимо разработать консольное приложение WarehouseRequest Console для учета материалов на складе и обработки заявок на их выдачу сотрудникам организации.

В компании есть внутренний склад, на котором хранятся различные материалы: бумага, картриджи, кабели, канцелярские товары, расходники, комплектующие и другие позиции. Сотрудники разных отделов могут создавать заявки на получение материалов. Кладовщик должен видеть остатки на складе, обрабатывать заявки, пополнять материалы и списывать их после выдачи.

Приложение должно позволять вести учет материалов, создавать заявки на выдачу, контролировать остатки, изменять статусы заявок и получать краткую статистику.

## Цель работы:
Создать консольную программу, которая позволяет добавлять материалы на склад, просматривать остатки, создавать и обрабатывать заявки на выдачу материалов, а также сохранять данные между запусками программы.

## Функциональные требования:

### 1. Главное меню

При запуске программы пользователь должен видеть меню:

```txt
=== WarehouseRequest Console ===

1. Добавить материал на склад
2. Показать все материалы
3. Создать заявку на выдачу
4. Показать все заявки
5. Найти заявку
6. Изменить статус заявки
7. Пополнить остаток материала
8. Списать материал по заявке
9. Показать статистику
10. Сохранить данные
0. Выход
```

Пользователь выбирает действие путем ввода номера пункта меню.

### 2. Добавление материала на склад

Программа должна позволять добавить новый материал на склад.

При добавлении материала пользователь вводит следующие данные:

```txt
- название материала
- категория материала
- единица измерения
- количество на складе
- минимальный остаток
```

**Категория материала может быть любой, например:**

- канцелярия
- расходники
- техника
- комплектующие
- упаковка
- другое

**Единица измерения может быть любой, например:**

- шт.
- пачка
- коробка
- метр
- кг

**Каждому материалу должен автоматически присваиваться уникальный номер.**

**Пример материала:**

```txt
Номер материала: 1
Название: Бумага A4
Категория: канцелярия
Единица измерения: пачка
Количество на складе: 25
Минимальный остаток: 5
```

### 3. Просмотр всех материалов

Программа должна выводить список всех материалов, которые есть на складе.

Для каждого материала необходимо отображать:

```txt
- номер материала
- название материала
- категория
- единица измерения
- количество на складе
- минимальный остаток
```

Если материалов нет, программа должна вывести сообщение:

`Материалы отсутствуют.`

Если количество материала меньше или равно минимальному остатку, рядом с материалом желательно выводить предупреждение:

`Требуется пополнение`

**Пример вывода:**

```txt
Номер материала: 2
Название: Картридж HP 106A
Категория: расходники
Единица измерения: шт.
Количество на складе: 2
Минимальный остаток: 3
Статус остатка: Требуется пополнение
```

### 4. Создание заявки на выдачу

Программа должна позволять создать заявку на выдачу материала сотруднику.

При создании заявки пользователь вводит:

```txt
- ФИО сотрудника
- отдел сотрудника
- номер материала
- количество материала
- причина выдачи
- статус заявки
```

**Статус заявки может принимать одно из значений:**

- новая
- одобрена
- отклонена
- выполнена

**Каждой заявке должен автоматически присваиваться уникальный номер.**

Перед созданием заявки программа должна проверить:

1. Существует ли материал с указанным номером.
2. Количество запрашиваемого материала должно быть больше 0.
3. Запрашиваемое количество не должно превышать количество материала на складе.

Если все проверки пройдены, заявка создается.

**Пример заявки:**

```txt
Номер заявки: 1
Сотрудник: Иванов Иван Иванович
Отдел: Бухгалтерия
Материал: Бумага A4
Количество: 3 пачка
Причина выдачи: Печать отчетных документов
Статус: новая
```

### 5. Просмотр всех заявок

Программа должна выводить список всех созданных заявок.

Для каждой заявки необходимо отображать:

```txt
- номер заявки
- ФИО сотрудника
- отдел
- название материала
- количество материала
- причина выдачи
- статус заявки
```

Если заявок нет, программа должна вывести сообщение:

`Заявки отсутствуют.`

### 6. Поиск заявки

Программа должна позволять искать заявки по разным параметрам.

Необходимо реализовать поиск:

```txt
- по ФИО сотрудника
- по отделу
- по названию материала
- по статусу заявки
```

При выборе пункта `Найти заявку` пользователь должен выбрать критерий поиска, затем ввести значение для поиска.

Пример меню поиска:

```txt
1. Поиск по сотруднику
2. Поиск по отделу
3. Поиск по материалу
4. Поиск по статусу
```

Если подходящие заявки найдены, программа выводит их список.

Если заявки не найдены, программа выводит сообщение:

`Заявки не найдены.`

### 7. Изменение статуса заявки

Программа должна позволять изменить статус существующей заявки.

Для изменения статуса пользователь вводит номер заявки, после чего выбирает новый статус:

```txt
1. новая
2. одобрена
3. отклонена
4. выполнена
```

Если заявка с указанным номером существует, программа изменяет ее статус и выводит сообщение:

`Статус заявки успешно изменен.`

Если заявка не найдена, программа выводит сообщение:

`Заявка с указанным номером не найдена.`

### 8. Пополнение остатка материала

Программа должна позволять увеличить количество материала на складе.

Для пополнения остатка пользователь вводит:

```txt
- номер материала
- количество для добавления
```

Если материал с указанным номером существует, программа увеличивает его остаток и выводит сообщение:

`Остаток материала успешно пополнен.`

Если материал не найден, программа выводит сообщение:

`Материал с указанным номером не найден.`

Если пользователь вводит отрицательное количество или ноль, программа должна вывести сообщение:

`Количество для пополнения должно быть больше нуля.`

### 9. Списание материала по заявке

Программа должна позволять списать материал со склада по заявке.

Для списания пользователь вводит номер заявки.

Списание возможно только если:

```txt
- заявка существует
- статус заявки: одобрена
- на складе достаточно материала
```

После успешного списания:

```txt
- количество материала на складе уменьшается
- статус заявки меняется на выполнена
```

Если заявка не найдена, программа выводит сообщение:

`Заявка с указанным номером не найдена.`

Если заявка не одобрена, программа выводит сообщение:

`Списание возможно только по одобренной заявке.`

Если материала на складе недостаточно, программа выводит сообщение:

`Недостаточно материала на складе.`

Если списание выполнено успешно, программа выводит сообщение:

`Материал успешно списан по заявке.`

### 10. Статистика

Программа должна выводить краткую статистику:

```txt
- общее количество материалов
- общее количество заявок
- количество новых заявок
- количество одобренных заявок
- количество отклоненных заявок
- количество выполненных заявок
- количество материалов, требующих пополнения
- общее количество единиц материалов на складе
```

Пример вывода:

```txt
=== Статистика ===

Всего материалов: 8
Всего заявок: 15
Новые заявки: 4
Одобренные заявки: 3
Отклоненные заявки: 2
Выполненные заявки: 6
Материалы, требующие пополнения: 2
Всего единиц материалов на складе: 147
```

### 11. Сохранение данных

Программа должна сохранять данные о материалах и заявках в файл.

Формат файла можно выбрать самостоятельно:

- txt
- csv
- json

После сохранения программа должна вывести сообщение:

`Данные успешно сохранены.`

### 12. Загрузка данных

При запуске программы необходимо загружать ранее сохраненные материалы и заявки из файла, если такой файл существует.

Если файл отсутствует, программа начинает работу с пустыми списками материалов и заявок.

### 13. Выход из программы

При выборе пункта `0. Выход` программа должна завершить работу.

Перед выходом желательно выполнить автоматическое сохранение данных в файл.

8. Если запрашиваемое количество превышает остаток на складе, необходимо вывести сообщение:

10. Выполненная заявка не должна списываться повторно.
11. Отклоненная заявка не должна списываться.
12. Данные должны сохраняться между запусками программы.

1. Пользователь запускает программу.

## Результат выполнения работы:

В результате должно быть создано консольное приложение WarehouseRequest Console, которое позволяет вести учет складских материалов, создавать заявки на их выдачу, контролировать остатки, списывать материалы по заявкам и сохранять данные между запусками программы.

````python
from dataclasses import dataclass
from typing import List, Optional, Dict
import json
import csv
import os

@dataclass
class Material:
    id: int
    name: str
    category: str
    unit: str
    quantity: int
    min_quantity: int

@dataclass
class Request:
    id: int
    employee_name: str
    department: str
    material_id: int
    quantity: int
    reason: str
    status: str

class WarehouseManager:
    def __init__(self, data_file: str = "warehouse_data"):
        self.materials: List[Material] = []
        self.requests: List[Request] = []
        self.next_material_id = 1
        self.next_request_id = 1
        self.data_file = data_file
        self.load_data()

    def save_data(self, format: str = "json"):
        if format == "json":
            self._save_json()
        elif format == "csv":
            self._save_csv()
        elif format == "txt":
            self._save_txt()
        else:
            print("Неподдерживаемый формат. Используйте: json, csv, txt")
            return

        print(f"Данные успешно сохранены в формате {format}.")

    def _save_json(self):
        data = {
            "materials": [m.__dict__ for m in self.materials],
            "requests": [r.__dict__ for r in self.requests],
            "next_material_id": self.next_material_id,
            "next_request_id": self.next_request_id
        }

        with open(f"{self.data_file}.json", 'w', encoding='utf-8') as f:
            json.dump(data, f, ensure_ascii=False, indent=2)

    def _save_csv(self):
        with open(f"{self.data_file}_materials.csv", 'w', newline='', encoding='utf-8') as f:
            writer = csv.writer(f)
            writer.writerow(["ID", "Название", "Категория", "Единица", "Количество", "Мин. остаток"])
            for m in self.materials:
                writer.writerow([m.id, m.name, m.category, m.unit, m.quantity, m.min_quantity])

        with open(f"{self.data_file}_requests.csv", 'w', newline='', encoding='utf-8') as f:
            writer = csv.writer(f)
            writer.writerow(["ID", "Сотрудник", "Отдел", "ID материала", "Количество", "Причина", "Статус"])
            for r in self.requests:
                writer.writerow([r.id, r.employee_name, r.department, r.material_id, r.quantity, r.reason, r.status])

    def _save_txt(self):
        with open(f"{self.data_file}.txt", 'w', encoding='utf-8') as f:
            f.write("=== ДАННЫЕ СКЛАДА ===\n\n")

            f.write("МАТЕРИАЛЫ:\n")
            if not self.materials:
                f.write("Материалы отсутствуют.\n")
            else:
                for m in self.materials:
                    f.write(f"ID: {m.id}\n")
            f.write(f"Название: {m.name}\n")
            f.write(f"Категория: {m.category}\n")
            f.write(f"Единица: {m.unit}\n")
            f.write(f"Количество: {m.quantity}\n")
            f.write(f"Мин. остаток: {m.min_quantity}\n")
            if m.quantity <= m.min_quantity:
                f.write("Статус: Требуется пополнение\n")
            f.write("- " * 30 + "\n")

            f.write("\nЗАЯВКИ:\n")
            if not self.requests:
                f.write("Заявки отсутствуют.\n")
            else:
                for r in self.requests:
                    material = self._find_material(r.material_id)
                    material_name = material.name if material else "Неизвестный"
            f.write(f"ID заявки: {r.id}\n")
            f.write(f"Сотрудник: {r.employee_name}\n")
            f.write(f"Отдел: {r.department}\n")
            f.write(f"Материал: {material_name}\n")
            f.write(f"Количество: {r.quantity}\n")
            f.write(f"Причина: {r.reason}\n")
            f.write(f"Статус: {r.status}\n")
            f.write("- " * 30 + "\n")

            f.write(f"\nМЕТАДАННЫЕ:\n")
            f.write(f"Следующий ID материала: {self.next_material_id}\n")
            f.write(f"Следующий ID заявки: {self.next_request_id}\n")

    def load_data(self):
        if os.path.exists(f"{self.data_file}.json"):
            self._load_json()
        elif os.path.exists(f"{self.data_file}_materials.csv") and os.path.exists(f"{self.data_file}_requests.csv"):
            self._load_csv()
        elif os.path.exists(f"{self.data_file}.txt"):
            self._load_txt()

    def _load_json(self):
        try:
            with open(f"{self.data_file}.json", 'r', encoding='utf-8') as f:
                data = json.load(f)

            self.materials = [Material(**mat) for mat in data.get("materials", [])]
            self.requests = [Request(**req) for req in data.get("requests", [])]
            self.next_material_id = data.get("next_material_id", 1)
            self.next_request_id = data.get("next_request_id", 1)
        except (json.JSONDecodeError, KeyError) as e:
            print(f"Ошибка при загрузке JSON-данных: {e}")

    def _load_csv(self):
        try:
            with open(f"{self.data_file}_materials.csv", 'r', newline='', encoding='utf-8') as f:
                reader = csv.DictReader(f)
                for row in reader:
                    self.materials.append(Material(
                id=int(row["ID"]),
                name=row["Название"],
                category=row["Категория"],
                unit=row["Единица"],
                quantity=int(row["Количество"]),
                min_quantity=int(row["Мин. остаток"])
            ))

            with open(f"{self.data_file}_requests.csv", 'r', newline='', encoding='utf-8') as f:
                reader = csv.DictReader(f)
                for row in reader:
                    self.requests.append(Request(
                id=int(row["ID"]),
                employee_name=row["Сотрудник"],
                department=row["Отдел"],
                material_id=int(row["ID материала"]),
                quantity=int(row["Количество"]),
                reason=row["Причина"],
                status=row["Статус"]
            ))
        except Exception as e:
            print(f"Ошибка при загрузке CSV-данных: {e}")

    def _load_txt(self):
        print("Загрузка из TXT-формата не реализована (требует парсинга текста).")

    def add_material(self):
        print("\n--- Добавление материала ---")
        name = input("Название материала: ")
        category = input("Категория материала: ")
        unit = input("Единица измерения: ")

        try:
            quantity = int(input("Количество на складе: "))
            if quantity < 0:
                print("Количество материала не может быть отрицательным.")
                return

            min_quantity = int(input("Минимальный остаток: "))
            if min_quantity < 0:
                print("Минимальный остаток не может быть отрицательным.")
                return
        except ValueError:
            print("Некорректный ввод количества. Используйте целые числа.")
            return

        material = Material(
            id=self.next_material_id,
            name=name,
            category=category,
            unit=unit,
            quantity=quantity,
            min_quantity=min_quantity
        )
        self.materials.append(material)
        self.next_material_id += 1
        print(f"Материал '{name}' успешно добавлен с номером {material.id}.")

    def show_materials(self):
        if not self.materials:
            print("Материалы отсутствуют.")
            return

        for material in self.materials:
            print(f"\nНомер материала: {material.id}")
            print(f"Название: {material.name}")
            print(f"Категория: {material.category}")
            print(f"Единица измерения: {material.unit}")
            print(f"Количество на складе: {material.quantity}")
            print(f"Минимальный остаток: {material.min_quantity}")

            if material.quantity <= material.min_quantity:
                print("Статус остатка: Требуется пополнение")

    def create_request(self):
        print("\n--- Создание заявки на выдачу ---")
        employee_name = input("ФИО сотрудника: ")
        department = input("Отдел сотрудника: ")

        try:
            material_id = int(input("Номер материала: "))
            material = self._find_material(material_id)
            if not material:
                print("Материал с указанным номером не найден.")
                return

            request_quantity = int(input("Количество материала: "))
            if request_quantity <= 0:
                print("Количество в заявке должно быть больше нуля.")
                return
            if request_quantity > material.quantity:
                print("Запрашиваемое количество превышает остаток на складе.")
                return

            reason = input("Причина выдачи: ")
            status = input("Статус заявки (новая/одобрена/отклонена/выполнена): ")

            request = Request(
                id=self.next_request_id,
                employee_name=employee_name,
                department=department,
                material_id=material_id,
                quantity=request_quantity,
                reason=reason,
                status=status
            )
            self.requests.append(request)
            self.next_request_id += 1
            print(f"Заявка успешно создана с номером {request.id}.")
        except ValueError:
            print("Некорректный ввод номера материала или количества.")

    def show_requests(self):
        if not self.requests:
            print("Заявки отсутствуют.")
            return

        for request in self.requests:
            material = self._find_material(request.material_id)
            material_name = material.name if material else "Неизвестный материал"
            print(f"\nНомер заявки: {request.id}")
            print(f"Сотрудник: {request.employee_name}")
            print(f"Отдел: {request.department}")
            print(f"Материал: {material_name}")
            print(f"Количество: {request.quantity} {material.unit if material else ''}")
            print(f"Причина выдачи: {request.reason}")
            print(f"Статус: {request.status}")

    def search_requests(self):
        print("\n--- Поиск заявок ---")
        print("1. Поиск по сотруднику")
        print("2. Поиск по отделу")
        print("3. Поиск по материалу")
        print("4. Поиск по статусу")
        choice = input("Выберите критерий поиска: ")
        search_value = input("Введите значение для поиска: ").lower()

        found_requests = []

        if choice == "1":
            found_requests = [r for r in self.requests if search_value in r.employee_name.lower()]
        elif choice == "2":
            found_requests = [r for r in self.requests if search_value in r.department.lower()]
        elif choice == "3":
            matching_materials = [m for m in self.materials if search_value in m.name.lower()]
            for mat in matching_materials:
                found_requests.extend([r for r in self.requests if r.material_id == mat.id])
        elif choice == "4":
            found_requests = [r for r in self.requests if search_value in r.status.lower()]
        else:
            print("Некорректный выбор критерия поиска.")
            return
        
        if not found_requests:
            print("Заявки не найдены.")
        else:
            for request in found_requests:
                material = self._find_material(request.material_id)
                material_name = material.name if material else "Неизвестный материал"
                print(f"\nНомер заявки: {request.id}")
                print(f"Сотрудник: {request.employee_name}")
                print(f"Отдел: {request.department}")
                print(f"Материал: {material_name}")
                print(f"Количество: {request.quantity} {material.unit if material else ''}")
                print(f"Причина выдачи: {request.reason}")
                print(f"Статус: {request.status}")

    def change_request_status(self):
        print("\n--- Изменение статуса заявки ---")
        try:
            request_id = int(input("Введите номер заявки: "))
            request = self._find_request(request_id)
            if not request:
                print("Заявка с указанным номером не найдена.")
                return

            print("Выберите новый статус:")
            print("1. новая")
            print("2. одобрена")
            print("3. отклонена")
            print("4. выполнена")
            choice = input("Ваш выбор: ")

            new_status = {
                "1": "новая",
                "2": "одобрена",
                "3": "отклонена",
                "4": "выполнена"
            }.get(choice)

            if new_status is None:
                print("Недопустимый статус. Повторите ввод.")
                return

            request.status = new_status
            print("Статус заявки успешно изменён.")
        except ValueError:
            print("Некорректный ввод номера заявки.")

    def replenish_material(self):
        print("\n--- Пополнение остатка материала ---")
        try:
            material_id = int(input("Введите номер материала: "))
            material = self._find_material(material_id)
            if not material:
                print("Материал с указанным номером не найден.")
                return

            add_quantity = int(input("Введите количество для добавления: "))
            if add_quantity <= 0:
                print("Количество для пополнения должно быть больше нуля.")
                return

            material.quantity += add_quantity
            print("Остаток материала успешно пополнен.")
        except ValueError:
            print("Некорректный ввод номера материала или количества.")

    def write_off_material(self):
        print("\n--- Списание материала по заявке ---")
        try:
            request_id = int(input("Введите номер заявки: "))
            request = self._find_request(request_id)
            if not request:
                print("Заявка с указанным номером не найдена.")
                return

            if request.status != "одобрена":
                print("Списание возможно только по одобренной заявке.")
                return
            
            material = self._find_material(request.material_id)
            if not material or material.quantity < request.quantity:
                print("Недостаточно материала на складе.")
                return

            material.quantity -= request.quantity
            request.status = "выполнена"
            print("Материал успешно списан по заявке.")
        except ValueError:
            print("Некорректный ввод номера заявки.")

    def show_statistics(self):
        print("\n=== Статистика ===")
        print(f"Всего материалов: {len(self.materials)}")
        print(f"Всего заявок: {len(self.requests)}")
        print(f"Новые заявки: {sum(1 for r in self.requests if r.status == 'новая')}")
        print(f"Одобренные заявки: {sum(1 for r in self.requests if r.status == 'одобрена')}")
        print(f"Отклоненные заявки: {sum(1 for r in self.requests if r.status == 'отклонена')}")
        print(f"Выполненные заявки: {sum(1 for r in self.requests if r.status == 'выполнена')}")
        print(f"Материалы, требующие пополнения: {sum(1 for m in self.materials if m.quantity <= m.min_quantity)}")
        print(f"Всего единиц материалов на складе: {sum(m.quantity for m in self.materials)}")

    def _find_material(self, material_id: int) -> Optional[Material]:
        return next((m for m in self.materials if m.id == material_id), None)

    def _find_request(self, request_id: int) -> Optional[Request]:
        return next((r for r in self.requests if r.id == request_id), None)

class WarehouseApp:
    def __init__(self):
        self.warehouse = WarehouseManager()

    def show_menu(self):
        print("\n=== WarehouseRequest Console ===")
        print("1. Добавить материал на склад")
        print("2. Показать все материалы")
        print("3. Создать заявку на выдачу")
        print("4. Показать все заявки")
        print("5. Найти заявку")
        print("6. Изменить статус заявки")
        print("7. Пополнить остаток материала")
        print("8. Списать материал по заявке")
        print("9. Показать статистику")
        print("10. Сохранить данные (JSON)")
        print("11. Сохранить данные (CSV)")
        print("12. Сохранить данные (TXT)")
        print("0. Выход")
        print("Выберите действие: ", end="")

    def run(self):
        while True:
            self.show_menu()
            choice = input().strip()

            match choice:
                case "1":
                    self.warehouse.add_material()
                case "2":
                    self.warehouse.show_materials()
                case "3":
                    self.warehouse.create_request()
                case "4":
                    self.warehouse.show_requests()
                case "5":
                    self.warehouse.search_requests()
                case "6":
                    self.warehouse.change_request_status()
                case "7":
                    self.warehouse.replenish_material()
                case "8":
                    self.warehouse.write_off_material()
                case "9":
                    self.warehouse.show_statistics()
                case "10":
                    self.warehouse.save_data("json")
                case "11":
                    self.warehouse.save_data("csv")
                case "12":
                    self.warehouse.save_data("txt")
                case "0":
                    self.warehouse.save_data("json")
                    print("Выход из программы.")
                    break
                case _:
                    print("Некорректный выбор. Попробуйте снова.")

def main():
    app = WarehouseApp()
    app.run()

main()
````
